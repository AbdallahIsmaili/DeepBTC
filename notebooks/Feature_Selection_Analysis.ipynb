# Feature Selection Analysis

Analyze and select best features for model performance

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier

sns.set_style('whitegrid')
```

## 1. Correlation Analysis

```python
# Load data
import kagglehub
path = kagglehub.dataset_download('oussamataghlaoui/btc-oracle-on-chain-sentiment-and-macro-data')
import glob
csv_files = glob.glob(path + '/*.csv')
df = pd.read_csv(csv_files[0], parse_dates=['Datetime'], index_col='Datetime')

# Select numeric columns
numeric_cols = df.select_dtypes(include=[np.number]).columns
print(f'Numeric features: {len(numeric_cols)}')
```

```python
# Correlation matrix
corr_matrix = df[numeric_cols].corr()

# Find highly correlated pairs (>0.95)
high_corr = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        if abs(corr_matrix.iloc[i, j]) > 0.95:
            high_corr.append((corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j]))

print(f'Highly correlated pairs (>0.95): {len(high_corr)}')
```
